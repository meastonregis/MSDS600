### Initialize the API Client

```python
client = Client()  
```

```python
import authenticate
#Bluesky API looks for the Client_Id and an App Password.
#Client ID is the handle of the user using this API and looks like this:  username..bsky.social
#App Password is NOT the User's password. Instead it can be found under Settings - Privacy & Security - App Password
#The authenticate file only has the following two lines:

#client_id = 'username.bsky.social'
#client_app_pw = 'xxxx-xxxx-xxxx-xxxx'
```

```python
profile = client.login(authenticate.client_id, authenticate.client_app_pw)
print('Welcome,', profile.display_name)
```

### Read a personal Timeline

```python
#Read a Timeline - API Limits are listed as 300 posts per day. I have not tested this limit.
timeline = client.get_timeline(algorithm='reverse-chronological')
bsky_data = {'object': [], 'handle':[],'text':[]}
for feed_view in timeline.feed:
    bsky_data['object'].append(feed_view)
    bsky_data['handle'].append(feed_view.post.author.handle)
    bsky_data['text'].append(feed_view.post.record.text)
    #print(feed_view.post.author.handle, '-', feed_view.post.record.text)
```
